```python
import pandas as pd

import re

from collections import Counter



df = pd.read_csv("repubblica.csv")

df["token"] = df["full_text"].apply(lambda t: re.findall(r"\w+", str(t).lower()))

df.head()
```

```python
paesi = {

    "libia": ["libia", "libico", "libici", "libica", "libiche"],

    "sudan": ["sudan", "sudanese", "sudanesi"],

    "iraq": ["iraq", "iracheno", "iracheni", "irachena"],

    "afghanistan": ["afghanistan", "afghano", "afghani", "afghana"],

    "siria": ["siria", "siriano", "siriani", "siriana"],

}



def conta_paesi(tokens):

    c = Counter(tokens)

    return {p: sum(c[f] for f in forme) for p, forme in paesi.items()}



conteggi = df["token"].apply(conta_paesi).apply(pd.Series)

print(conteggi.sum())
```

```python
anni = [2003, 2014, 2015, 2016, 2025]

paesi_istat = ["Libia", "Sudan", "Iraq", "Afghanistan", "Siria"]



pezzi = []

for anno in anni:

    f = pd.read_csv(f"Dati_RCS/Dati_cittadinanza_{anno}.csv", sep=";")

    pezzi.append(f[f["Stato di cittadinanza"].isin(paesi_istat)])



istat = pd.concat(pezzi)

residenti = istat.groupby(["Anno", "Stato di cittadinanza"])["Totale"].sum().unstack()

print(residenti)

residenti.to_csv("residenti_istat.csv")

```

 Residenti stranieri per cittadinanza (ISTAT)

Viene mostrato il numero di residenti in Italia con cittadinanza libica, sudanese, irachena, afghana e siriana negli anni di indagine 2003, 2014, 2015, 2016 e 2025.



Nel 2003, prima delle crisi, le presenze sono contenute: la Siria (10300) e la Libia (8810) sono le più numerose, l'Iraq ha 4640 residenti, mentre Sudan (2845) e Afghanistan (2430) sono le meno rappresentate.



Tra il 2003 e il 2014, con l'inizio della crisi, l'Afghanistan cresce molto (da 2430 a 28750) e la Siria subisce di più di un raddoppio (da 10300 a 22385). L'Iraq passa da 4640 a 13385, mentre la Libia resta stabile.



Nel 2025 tutti i paesi aumentano e l'Afghanistan diventa il più numeroso, con 88975 residenti, seguito dall'Iraq (39690) e dalla Siria (34480).

```python
df = pd.read_csv("repubblica.csv")

print(df.shape)

df.head()
```

```python
import re



df["parole"] = df["full_text"].apply(lambda t: set(re.findall(r"\w+", str(t).lower())))



paesi = {

    "Libia": {"libia", "libico", "libica", "libici", "libiche"},

    "Sudan": {"sudan", "sudanese", "sudanesi"},

    "Iraq": {"iraq", "iracheno", "irachena", "iracheni", "irachene"},

    "Afghanistan": {"afghanistan", "afghano", "afghana", "afghani", "afghane"},

    "Siria": {"siria", "siriano", "siriana", "siriani", "siriane"},

}

sicurezza = {"ladro", "ladri", "furto", "furti", "rapina", "rapine", "rapitore",

             "rapitori", "rapimento", "criminale", "criminali", "criminalità",

             "reato", "reati", "violenza", "sicurezza"}



for p, termini in paesi.items():

    df[p] = df["parole"].apply(lambda s: bool(s & termini))

df["paese_citato"] = df[list(paesi)].any(axis=1)

df["sicurezza"] = df["parole"].apply(lambda s: bool(s & sicurezza))



print("Articoli per paese:\n", df[list(paesi)].sum())

print("Sicurezza in tutti gli articoli:", df["sicurezza"].mean())

print("Sicurezza negli articoli sui paesi:", df[df["paese_citato"]]["sicurezza"].mean())

```

Euristiche sugli articoli di Repubblica



Euristica 1: presenza dei paesi negli articoli. 
Per ogni articolo è stato controllato se contiene almeno una parola legata a uno dei cinque paesi. Gli articoli che citano ciascun paese sono: Siria (3292), Iraq (1976), Libia (1833), Afghanistan (1063) e Sudan (713). Siria e Iraq sono quindi i paesi più presenti nel giornale, mentre il Sudan è il meno citato.



Euristica 2: campo semantico della sicurezza.
È stato definito un insieme di parole legate alla sicurezza e alla criminalità (ad esempio "ladro", "furto", "rapina", "rapitore", "criminalità", "reato") e si è controllato in quanti articoli compaiono queste terminologie. Questi termini sono presenti nel 25% circa di tutti gli articoli, mentre tra quelli che citano almeno uno dei cinque paesi la quota sale al 47% circa.

```python
print("Per paese (parole di sicurezza):")

for p in paesi:

    print(p, round(df[df[p]]["sicurezza"].mean(), 3))



stretta = sicurezza - {"violenza", "sicurezza"}

df["sicurezza_stretta"] = df["parole"].apply(lambda s: bool(s & stretta))



print("\nCon lista ristretta (senza 'violenza' e 'sicurezza'):")

print("Tutti:", round(df["sicurezza_stretta"].mean(), 3))

print("Paesi:", round(df[df["paese_citato"]]["sicurezza_stretta"].mean(), 3))

for p in paesi:

    print(p, round(df[df[p]]["sicurezza_stretta"].mean(), 3))
```

Confronto per singolo paese.
 Con la lista completa, la quota di articoli con parole di sicurezza è simile per tutti e cinque i paesi, tra il 48% e il 50% (Libia 50%, Iraq 49%, Sudan 48,5%, Afghanistan 48%, Siria 48%), contro il 25% circa dell'intero giornale. Nessun paese si distingue in modo netto dagli altri.



Lista ristretta
 Abbiamo poi ripetuto il conteggio togliendo i termini più generici ("violenza" e "sicurezza"), che compaiono spesso anche in articoli su guerre e terrorismo nei paesi di origine. Con la lista ristretta la differenza si riduce, infatti i termini legati alla criminalità compaiono nel 9,2% di tutti gli articoli e nel 12,6% di quelli che citano uno dei cinque paesi. Per singolo paese i valori sono Libia 17%, Afghanistan 13,2%, Sudan 11,9%, Siria 11,8% e Iraq 10,9%. 



Dunque, l'associazione tra i paesi considerati e il campo semantico della sicurezza esiste, ma è molto più debole di quanto sembrasse con la lista completa: gran parte del divario iniziale dipendeva da parole generiche legate ai conflitti. Il risultato dipende anche dalle liste di parole scelte e misura solo la co-occorrenza nello stesso articolo, non il modo in cui il giornale collega i due temi.
